# 03. Performance Prediction Model Training
Multiclass classification for construction performance score (Poor, Average, Good, Excellent).

```python
import sys
sys.path.append('../src')
from conark.data.loader import DataLoader
from conark.features.engineering import create_feature_pipeline
from conark.data.splitter import time_series_split
from conark.training.train_performance import train_performance_model

loader = DataLoader()
df_raw = loader.load_raw_dataset()
df_feat = create_feature_pipeline(df_raw)
train_df, val_df, test_df = time_series_split(df_feat)
model, meta = train_performance_model(train_df, val_df)
print('Trained Performance Model Metadata:', meta)
```
